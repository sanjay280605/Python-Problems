```python
#Q1.Write a Python program to print the first digit of a given number
n = int(input("Enter a number: "))

while n >= 10:
    n //= 10

print("First digit =", n)
```

```python
#Q2.Write a Python program to check whether the first digit of a given number is odd or even.
n = int(input("Enter a number: "))

while n >= 10:
    n //= 10

if n % 2 == 0:
    print("First digit is Even")
else:
    print("First digit is Odd")
```

```python
#Q3.Write a Python program to check whether a given digit occurs in a number.
n = int(input("Enter a number: "))
digit = int(input("Enter digit to search: "))

temp = n
found = False

while temp > 0:
    if temp % 10 == digit:
        found = True
        break
    temp //= 10

if found:
    print("Digit Found")
else:
    print("Digit Not Found")
```

```python
#Q4.Write a Python program to count the occurrence of a digit in a number.
n = int(input("Enter a number: "))
digit = int(input("Enter digit: "))

count = 0
temp = n

while temp > 0:
    if temp % 10 == digit:
        count += 1
    temp //= 10

print("Occurrence =", count)
```

```python
#Q5.Write a Python program to check whether the digits of a number are in ascending order.
n = input("Enter a number: ")

if list(n) == sorted(n):
    print("Digits are in Ascending Order")
else:
    print("Digits are Not in Ascending Order")
```

```python
#Q6.Write a Python program to check whether the digits of a number are in descending order.
n = input("Enter a number: ")

if list(n) == sorted(n, reverse=True):
    print("Digits are in Descending Order")
else:
    print("Digits are Not in Descending Order")
```

```python
#Q7.Write a Python program to swap the first and last digit of a given number.
n = input("Enter a number: ")

if len(n) == 1:
    print("After swapping:", n)
else:
    new_num = n[-1] + n[1:-1] + n[0]
    print("After swapping:", new_num)
```

```python
#Q8.Write a Python program to find the frequency of each digit in a given number.
n = input("Enter a number: ")

for i in range(10):
    print(i, "occurs", n.count(str(i)), "times")
```
